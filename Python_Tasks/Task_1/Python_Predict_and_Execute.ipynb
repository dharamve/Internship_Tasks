## Q35
**Expected output:**
```text
[[1, 99], [3, 4]]
```

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)

# Explanation: A shallow copy creates a new outer list but shares inner lists, so changing an inner element is visible through a.
````

## Q36
**Expected output:**
```text
[[1, 2], [3, 4]]
```

````python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)

# Explanation: deepcopy creates independent nested lists, so editing b does not affect a.
````

## Quick revision
- `{}` creates an empty dictionary; `set()` creates an empty set.
- A tuple with one element needs a comma, for example `(25,)`.
- `==` compares values; `is` compares object identity.
- List slicing and shallow copying create a new outer list.
- `copy.deepcopy()` also copies nested lists.
- `*args` collects positional arguments in a tuple.
- `**kwargs` collects keyword arguments in a dictionary.
- `print()` displays output; `return` sends a value back to the caller.
