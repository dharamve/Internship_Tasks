## Q1. Find the Shortest Word in a Sentence

**Question:** Write a Python program to find the shortest word in a given sentence.

**Observe:** Input is `Python makes coding easy`. Expected output in the PDF: `easy`.

**Understand:** Identify what value or collection must be returned, including the tie-breaking and case-handling rules.

**Break:** Split the sentence into words, start with the first word as the shortest, then compare each remaining word's length. Update only when a strictly shorter word is found, so ties keep the first word.

**Logic (plain English):** Split the sentence into words, start with the first word as the shortest, then compare each remaining word's length. Update only when a strictly shorter word is found, so ties keep the first word.

**Test plan:** Run the sample from the PDF and a second test case to check the behavior on another input.

```python
def shortest_word(sentence):
    words = sentence.split()
    if not words:
        return None

    shortest = words[0]
    for word in words[1:]:
        if len(word) < len(shortest):
            shortest = word
    return shortest

sentence = "Python makes coding easy"
print(shortest_word(sentence))  # Sample test

print(shortest_word("I enjoy Python"))  # Additional test
```

## Q8. Find the Most Frequent Character

**Question:** Find the most frequent character in a string. If there is a tie, return the first one encountered. Ignore spaces.

**Observe:** Input is `banana`. Expected output in the PDF: `a`.

**Understand:** Identify what value or collection must be returned, including the tie-breaking and case-handling rules.

**Break:** Remove spaces. Count each character's occurrences, then scan the cleaned string in order and keep the first character with the highest count.

**Logic (plain English):** Remove spaces. Count each character's occurrences, then scan the cleaned string in order and keep the first character with the highest count.

**Test plan:** Run the sample from the PDF and a second test case to check the behavior on another input.

```python
def most_frequent_character(text):
    characters = [character for character in text if not character.isspace()]
    if not characters:
        return None

    best_character = characters[0]
    best_count = characters.count(best_character)
    for character in characters:
        frequency = characters.count(character)
        if frequency > best_count:
            best_character = character
            best_count = frequency
    return best_character

print(most_frequent_character("banana"))  # Sample test
print(most_frequent_character("a b c"))  # Additional test
```

```python
def repeated_characters(text):
    characters = [character for character in text if not character.isspace()]
    repeated = []
    for character in characters:
        if characters.count(character) > 1 and character not in repeated:
            repeated.append(character)
    return repeated

print(repeated_characters("programming"))  # Sample test
print(repeated_characters("a a b b c"))  # Additional test
```

## Q15. Create a Dictionary of Numbers and Their Even/Odd Status

**Logic (plain English):** For each number, check number % 2. A remainder of zero means Even; otherwise it is Odd.

**Test plan:** Run the sample from the PDF and a second test case to check the behavior on another input.

```python
def even_odd_dictionary(numbers):
    result = {}
    for number in numbers:
        if number % 2 == 0:
            result[number] = "Even"
        else:
            result[number] = "Odd"
    return result

print(even_odd_dictionary([1, 2, 3, 4, 5]))  # Sample test
print(even_odd_dictionary([0, -3, 8]))  # Additional test
```

## Q20. Find the Longest Word That Appears Only Once

**Question:** Find the longest word that appears exactly once. If qualifying words tie in length, return the first one. Treat uppercase and lowercase as the same.

**Observe:** Input is `Python makes coding useful Python`. Expected output in the PDF: `coding`.

**Understand:** Identify what value or collection must be returned, including the tie-breaking and case-handling rules.

**Break:** Convert words to lowercase for counting. Scan in original order, considering only words that occur once, and update the answer only when a strictly longer word is found.

**Logic (plain English):** Convert words to lowercase for counting. Scan in original order, considering only words that occur once, and update the answer only when a strictly longer word is found.

**Test plan:** Run the sample from the PDF and a second test case to check the behavior on another input.

```python
def longest_unique_word(sentence):
    words = sentence.lower().split()
    longest = None
    for word in words:
        if words.count(word) == 1:
            if longest is None or len(word) > len(longest):
                longest = word
    return longest

print(longest_unique_word("Python makes coding useful Python"))  # Sample test
print(longest_unique_word("cat elephant tiger"))  # Additional test
```
