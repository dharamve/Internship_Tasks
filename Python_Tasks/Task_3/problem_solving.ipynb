## Q1. Find the frequency of each word in a sentence

### Pseudocode

1. Create an empty dictionary.
2. Split the sentence into words and make each word lowercase.
3. Increase that word's count each time it appears.
4. Return the dictionary.

### Expected output

```text
{'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}
{'data': 2, 'helps': 1, 'teams': 1}
```

````python
def word_frequency(sentence):
    counts = {}
    for word in sentence.split():
        word = word.lower()
        counts[word] = counts.get(word, 0) + 1
    return counts

print(word_frequency("python is easy and python is powerful"))
print(word_frequency("data helps data teams"))
````

## Q2. Find the frequency of each character in a word

### Pseudocode

1. Start with an empty dictionary.
2. Visit each character in the word.
3. Add one to its count, starting at zero if it is new.
4. Return the character-count dictionary.

### Expected output

```text
{'b': 1, 'a': 3, 'n': 2}
{'h': 1, 'e': 1, 'l': 2, 'o': 1}
```

````python
def character_frequency(word):
    counts = {}
    for character in word:
        counts[character] = counts.get(character, 0) + 1
    return counts

print(character_frequency("banana"))
print(character_frequency("hello"))
````

## Q3. Find the first non-repeating character

### Pseudocode

1. Count how often every character appears.
2. Read the word from left to right.
3. Return the first character whose count is one.
4. If none qualifies, return None.

### Expected output

```text
w
None
```

````python
def first_non_repeating_character(word):
    counts = {}
    for character in word:
        counts[character] = counts.get(character, 0) + 1
    for character in word:
        if counts[character] == 1:
            return character
    return None

print(first_non_repeating_character("swiss"))
print(first_non_repeating_character("aabb"))
````

## Q4. Check whether two strings are anagrams

### Pseudocode

1. Remove spaces and convert both strings to lowercase.
2. Count the characters in each string.
3. Compare the two frequency dictionaries.
4. Return True if they match, otherwise False.

### Expected output

```text
True
False
```

````python
def are_anagrams(s1, s2):
    def counts(text):
        result = {}
        for character in text.replace(" ", "").lower():
            result[character] = result.get(character, 0) + 1
        return result
    return counts(s1) == counts(s2)

print(are_anagrams("listen", "silent"))
print(are_anagrams("hello", "world"))
````

## Q5. Find the first repeating character

### Pseudocode

1. Create an empty set to remember characters already seen.
2. Scan the string from left to right.
3. If the current character is already in the set, return it.
4. Otherwise add it to the set; return None if no repeat is found.

### Expected output

```text
r
None
```

````python
def first_repeating_character(text):
    seen = set()
    for character in text:
        if character in seen:
            return character
        seen.add(character)
    return None

print(first_repeating_character("programming"))
print(first_repeating_character("abcdef"))
````

## Q6. Move all zeros to the end of a list

### Pseudocode

1. Collect the non-zero values in their original order.
2. Collect all zero values.
3. Join the two groups and return the new list.

### Expected output

```text
[3, 1, 2, 4, 0, 0, 0]
[5, 0, 0]
```

````python
def move_zeros_to_end(nums):
    non_zeros = [number for number in nums if number != 0]
    zeros = [number for number in nums if number == 0]
    return non_zeros + zeros

print(move_zeros_to_end([0, 3, 0, 1, 2, 0, 4]))
print(move_zeros_to_end([0, 0, 5]))
````

## Q7. Find the second highest distinct number

### Pseudocode

1. Remove duplicate values from the list.
2. Sort the distinct values from largest to smallest.
3. If there are at least two values, return the second one.
4. Otherwise return None.

### Expected output

```text
30
None
```

````python
def second_highest_distinct(nums):
    distinct = sorted(set(nums), reverse=True)
    if len(distinct) < 2:
        return None
    return distinct[1]

print(second_highest_distinct([10, 20, 40, 30, 40, 20]))
print(second_highest_distinct([7, 7, 7]))
````

## Q8. Remove duplicates while preserving order

### Pseudocode

1. Make an empty result list and an empty set.
2. Visit each value in the input list.
3. Add it to the result only if it has not been seen before.
4. Return the result list.

### Expected output

```text
[4, 2, 5, 7]
[1, 2, 3]
```

````python
def remove_duplicates_preserve_order(nums):
    result = []
    seen = set()
    for number in nums:
        if number not in seen:
            result.append(number)
            seen.add(number)
    return result

print(remove_duplicates_preserve_order([4, 2, 4, 5, 2, 7, 5]))
print(remove_duplicates_preserve_order([1, 1, 2, 3, 2]))
````

## Q9. Find the missing number from 1 to N

### Pseudocode

1. Calculate the expected sum from 1 through N using N × (N + 1) ÷ 2.
2. Add all values currently in the list.
3. Subtract the list sum from the expected sum.
4. Return the difference.

### Expected output

```text
3
2
```

````python
def find_missing_number(nums, n):
    expected_sum = n * (n + 1) // 2
    return expected_sum - sum(nums)

print(find_missing_number([1, 2, 4, 5, 6], 6))
print(find_missing_number([1, 3, 4], 4))
````

## Q10. Find the first pair that sums to a target

### Pseudocode

1. Keep a set of values already visited.
2. For each number, calculate the value needed to reach the target.
3. If that needed value has already appeared, return the pair.
4. Otherwise remember the current number; return an empty list if no pair works.

### Expected output

```text
[2, 7]
[1, 4]
```

````python
def two_sum_first_pair(nums, target):
    seen = set()
    for number in nums:
        needed = target - number
        if needed in seen:
            return [needed, number]
        seen.add(number)
    return []

print(two_sum_first_pair([2, 7, 11, 15], 9))
print(two_sum_first_pair([1, 4, 2, 3], 5))
````

## Q11. Find the number closest to zero

### Pseudocode

1. Compare values by their absolute distance from zero.
2. When distances tie, prefer the positive value.
3. Return the value with the best comparison key.

### Expected output

```text
2
4
```

````python
def closest_to_zero(nums):
    return min(nums, key=lambda number: (abs(number), -number))

print(closest_to_zero([-8, -3, 2, 5, -2]))
print(closest_to_zero([-4, 4, 7]))
````

## Q12. Find common elements between two lists

### Pseudocode

1. Convert the second list to a set for quick membership checks.
2. Walk through the first list in its original order.
3. Add a value if it is in the second list and has not already been added.
4. Return the unique common values.

### Expected output

```text
[3, 4, 5]
[2, 3]
```

````python
def common_elements(list1, list2):
    second_values = set(list2)
    result = []
    seen = set()
    for value in list1:
        if value in second_values and value not in seen:
            result.append(value)
            seen.add(value)
    return result

print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))
print(common_elements([2, 2, 1, 3], [3, 2, 2]))
````

## Q13. Rotate a list to the right by K positions

### Pseudocode

1. If the list is empty, return an empty list.
2. Reduce K using the list length so large rotations are handled.
3. Split the list at the position that should move to the front.
4. Put the ending section before the starting section.

### Expected output

```text
[4, 5, 1, 2, 3]
[3, 1, 2]
```

````python
def rotate_right(nums, k):
    if not nums:
        return []
    k %= len(nums)
    if k == 0:
        return nums[:]
    return nums[-k:] + nums[:-k]

print(rotate_right([1, 2, 3, 4, 5], 2))
print(rotate_right([1, 2, 3], 4))
````

## Q14. Find the maximum sum of two consecutive elements

### Pseudocode

1. Start at the first position and examine each adjacent pair.
2. Add each value to the one immediately after it.
3. Keep track of the largest pair sum.
4. Return the largest sum.

### Expected output

```text
14
-7
```

````python
def max_consecutive_pair_sum(nums):
    return max(nums[index] + nums[index + 1]
               for index in range(len(nums) - 1))

print(max_consecutive_pair_sum([4, 7, 2, 9, 5]))
print(max_consecutive_pair_sum([-5, -2, -8]))
````

## Q15. Find all unique pairs that sum to a target

### Pseudocode

1. Examine distinct values and calculate each value's required partner.
2. Keep a pair only when its partner exists and the first value is not larger than the partner.
3. If both values are equal, make sure the input contains that value at least twice.
4. Store pairs uniquely, sort them, and return them as lists.

### Expected output

```text
[[1, 8], [2, 7], [4, 5]]
[[1, 3]]
```

````python
def all_unique_pairs(nums, target):
    values = set(nums)
    pairs = set()
    for number in values:
        partner = target - number
        if partner in values and number <= partner:
            if number != partner or nums.count(number) >= 2:
                pairs.add((number, partner))
    return [list(pair) for pair in sorted(pairs)]

print(all_unique_pairs([2, 4, 3, 5, 7, 8, 1], 9))
print(all_unique_pairs([1, 1, 2, 3, 4], 4))
````
